```python
# randomモジュールのrandint関数を取り込む
from random import randint
print('数当てゲームを始めます。3桁の数を当ててください！')

# 正解を作成
#answer = list()
#for n in range(3):
#   answer.append(randint(0, 9))

answer = [1,8,5]

is_continue = True
while is_continue == True:
    # 予想の入力
    prediction = list()
    for n in range(3):
        data = int(input(f'{n + 1}桁目の予想入力 (0~9) >>'))
        prediction.append(data)

    # 答え合わせ
    hit = 0
    blow = 0
    for n in range(3):
        if prediction[n] == answer[n]:
            hit += 1
        else:
            for m in range(3):
                if prediction[n] == answer[m]:
                    blow += 1
                    break

    # 結果発表
    print(f'{hit}ヒット！{blow}ボール！')
    if hit == 3:
        print('正解です！')
        is_continue = False
    else:
        if int(input('続けますか？ 1：続ける 2：終了 >>')) == 2:
            print(f'正解は{answer[0]}{answer[1]}{answer[2]}でした')
            is_continue = False
```

```python
import random

# (1) メッセージを表示
print("数当てゲームを始めます。3桁の数を当ててください！")

# (2) 0〜9のランダムな整数を3つ格納したリストanswerを準備
#answer = [random.randint(0, 9) for _ in range(3)]

answer = [1,8,5]

while True:
    prediction = []

    # (3) 1〜3桁目の予想を入力し、リストpredictionに格納
    for i in range(1, 4):
        num = int(input(f"{i}桁目の予想を入力 (0~9) >>"))
        prediction.append(num)

    hit = 0
    ball = 0

    # (4) 位置と数値の一致（ヒット）および位置が異なるが数値の一致（ボール）をカウント
    for i in range(3):
        if prediction[i] == answer[i]:
            hit += 1
        else:
            ball += 1
    ball -= hit # 新しいロジックを追加

    print(f"{hit}ヒット！{ball}ボール！")

    # (5) 3ヒットなら「正解です！」と表示して終了
    if hit == 3:
        print("正解です！")
        break

    # (6) それ以外の場合は継続の確認を表示
    continue_flag = int(input("続けますか？ 1：続ける 2：終了 >>"))

    # (7) 2が入力されたら正解を表示して終了（1の場合はループの先頭に戻る）
    if continue_flag == 2:
        print(f"正解は {answer} でした。ゲームを終了します。")
        break
```

```python
import random

print('数当てゲームを始めます。3桁の数を当ててください！')

# 正解の生成（リスト内包表記でシンプルに）
#answer = [random.randint(0, 9) for _ in range(3)]
answer = [1,8,5]
while True:
    # 予想の入力
    prediction = [int(input(f'{i + 1}桁目の予想入力 (0~9) >>')) for i in range(3)]

    # 判定処理
    hit = sum(p == a for p, a in zip(prediction, answer))
    # ヒットを除いた共通の数字をカウント（ボール）
    ball = sum(min(prediction.count(x), answer.count(x)) for x in set(prediction)) - hit

    print(f'{hit}ヒット！{ball}ボール！')

    if hit == 3:
        print('正解です！')
        break

    if int(input('続けますか？ 1：続ける 2：終了 >>')) == 2:
        print(f'正解は{"".join(map(str, answer))}でした')
        break
```
